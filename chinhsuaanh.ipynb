```python
import os
import cv2
import albumentations as A
from tqdm import tqdm

# Thư mục ảnh gốc và thư mục lưu ảnh augment
src_dir = "go_keo_B_va_C_duoi"
dst_dir = "go_keo_B_va_C_sau_khi_chinh_sua"
os.makedirs(dst_dir, exist_ok=True)
    
# Pipeline augment
transform = A.Compose([
    #A.HorizontalFlip(p=0.5),
    A.RandomBrightnessContrast(p=0.5),
    A.Rotate(limit=10, p=0.5),
    A.GaussianBlur(blur_limit=(3, 5), p=0.3),
    A.ShiftScaleRotate(shift_limit=0.05, scale_limit=0.05, rotate_limit=10, p=0.5),
])

# Mỗi ảnh → tạo thêm 1~3 bản augment
for img_name in tqdm(os.listdir(src_dir)):
    img_path = os.path.join(src_dir, img_name)
    img = cv2.imread(img_path)

    if img is None:
        print(f"❌ Không đọc được: {img_path}")
        continue

    base_name = os.path.splitext(img_name)[0]

    # Lưu ảnh gốc lại
    cv2.imwrite(os.path.join(dst_dir, f"{base_name}_original.jpg"), img)

    # Tạo thêm 2 ảnh augment
    for i in range(2):
        augmented = transform(image=img)['image']
        new_name = f"{base_name}_aug{i+1}.jpg"
        cv2.imwrite(os.path.join(dst_dir, new_name), augmented)

print("✅ Augmentation hoàn tất.")
```

```python
import cv2
import os
from pathlib import Path

# Define input directory
input_dir = r"C:\side project\data\Warp-A\train\images"
# Define target size (you can adjust these values)
target_width = 640
target_height = 640

def resize_images(input_directory, width, height):
    # Create Path object
    input_path = Path(input_directory)
    
    # Supported image extensions
    image_extensions = ['.jpg', '.jpeg', '.png', '.bmp']
    
    # Process each image in directory
    for img_path in input_path.iterdir():
        if img_path.suffix.lower() in image_extensions:
            try:
                # Read image
                img = cv2.imread(str(img_path))
                if img is None:
                    print(f"Error reading {img_path}")
                    continue
                
                # Resize image
                resized_img = cv2.resize(img, (width, height), interpolation=cv2.INTER_AREA)
                
                # Save resized image (overwriting original)
                cv2.imwrite(str(img_path), resized_img)
                print(f"Resized {img_path}")
                
            except Exception as e:
                print(f"Error processing {img_path}: {str(e)}")

# Execute the resize operation
resize_images(input_dir, target_width, target_height)
print("Resizing completed!")
```
