```python
import os
import shutil
import random

# Đường dẫn đến thư mục chứa ảnh
image_dir = "C:/side project/data/Warp-A/train/images"

# Các thư mục đích cho train, val, test
formatted_dir = "C:/side project/data/formatted_dataset"
subsets = ["train", "val", "test"]

# Tạo thư mục cho các tập con (nếu chưa có)
for subset in subsets:
    os.makedirs(os.path.join(formatted_dir, subset, "images"), exist_ok=True)
    os.makedirs(os.path.join(formatted_dir, subset, "labels"), exist_ok=True)

# Danh sách tất cả các ảnh trong thư mục
image_files = [f for f in os.listdir(image_dir) if f.endswith(('.jpg', '.png', '.jpeg'))]

# Xáo trộn danh sách ảnh để đảm bảo tính ngẫu nhiên
random.shuffle(image_files)

# Chia ảnh thành train, val, test (80%, 10%, 10%)
train_count = int(0.8 * len(image_files))
val_count = int(0.1 * len(image_files))

train_files = image_files[:train_count]
val_files = image_files[train_count:train_count + val_count]
test_files = image_files[train_count + val_count:]

# Hàm sao chép ảnh và file nhãn vào đúng thư mục
def copy_files(file_list, subset):
    for file in file_list:
        # Đường dẫn ảnh và file nhãn
        img_path = os.path.join(image_dir, file)
        label_path = img_path.replace("images", "labels").replace(".jpg", ".txt").replace(".png", ".txt").replace(".jpeg", ".txt")
        
        # Thư mục đích
        target_img_path = os.path.join(formatted_dir, subset, "images", file)
        target_label_path = os.path.join(formatted_dir, subset, "labels", file.replace(".jpg", ".txt").replace(".png", ".txt").replace(".jpeg", ".txt"))
        
        # Sao chép ảnh và nhãn
        shutil.copy(img_path, target_img_path)
        shutil.copy(label_path, target_label_path)

# Sao chép ảnh và nhãn vào thư mục train, val, test
copy_files(train_files, "train")
copy_files(val_files, "val")
copy_files(test_files, "test")

print("✅ Dữ liệu đã được chia và sao chép hoàn tất!")
```
