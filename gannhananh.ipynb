```python
import os
import cv2

# Đường dẫn đến dataset
dataset_path = "C:/side project/data/Warp-A/train"
image_dir = os.path.join(dataset_path, "images")
label_dir = os.path.join(dataset_path, "labels")

# Tạo thư mục labels nếu chưa tồn tại
os.makedirs(label_dir, exist_ok=True)

# Lấy danh sách ảnh (giữ nguyên thứ tự)
image_files = [f for f in os.listdir(image_dir) if f.endswith(('.jpg', '.png', '.jpeg'))]
total_images = len(image_files)

# Hàm xử lý khi click chuột
def get_mouse_coordinates(event, x, y, flags, param):
    global points, img, label_path

    if event == cv2.EVENT_LBUTTONDOWN:
        points.append((x, y))

        if len(points) == 2:
            x1, y1 = points[0]
            x2, y2 = points[1]
            xmin, xmax = min(x1, x2), max(x1, x2)
            ymin, ymax = min(y1, y2), max(y1, y2)

            # Vẽ bounding box
            cv2.rectangle(img, (xmin, ymin), (xmax, ymax), (0, 255, 0), 2)
            cv2.imshow("Image", img)

            # Tính toán theo định dạng YOLO
            img_height, img_width, _ = img.shape
            x_center = (xmin + xmax) / 2 / img_width
            y_center = (ymin + ymax) / 2 / img_height
            bbox_width = (xmax - xmin) / img_width
            bbox_height = (ymax - ymin) / img_height

            class_id = 1 # 👈 chỉnh class nếu cần

            # Ghi vào file .txt
            yolo_format = f"{class_id} {x_center:.6f} {y_center:.6f} {bbox_width:.6f} {bbox_height:.6f}\n"
            with open(label_path, "a") as f:
                f.write(yolo_format)

            print(f"✅ Ghi nhãn: {label_path} → {yolo_format.strip()}")
            points.clear()

# Gán nhãn từng ảnh
for i, image_file in enumerate(image_files, start=1):
    label_path = os.path.join(label_dir, os.path.splitext(image_file)[0] + ".txt")

    # 👉 Nếu ảnh đã có file nhãn, bỏ qua
    if os.path.exists(label_path):
        continue

    image_path = os.path.join(image_dir, image_file)
    img = cv2.imread(image_path)
    if img is None:
        print(f"❌ Lỗi đọc ảnh: {image_path}")
        continue

    print(f"({i}/{total_images}) Gán nhãn: {image_file}")
    points = []
    cv2.imshow("Image", img)
    cv2.setMouseCallback("Image", get_mouse_coordinates)
    cv2.waitKey(0)
    cv2.destroyAllWindows()

print("🎉 Gán nhãn hoàn tất!")
```
